# Get Dataset

```python
# Access API_key from txt file
api_key = ''

with open('api_key.txt') as input_file:
    
    for line in input_file:
        if(not line.startswith("#")):
        # Ignore lines that contain comments
            name, value = (
                item.strip() for item in line.split('=', 1)
            )
            if name=='api_key':
                
                api_key = value

print("This is your API Key: %s" % api_key)
```
